```python
import dpluspy
import pickle
import numpy as np
import sys
```

```python
numer = 0.0
denom = 0.0

for chrom in range(1, 23):
    mut_fname = f"../../mutation_maps/roulette_b37/roulette_b37_chr{chrom}.npy"
    bed_fname = f"../../bed_files/filterbed_1e-4M_exon_buffer/mask_chr{chrom}.bed.gz"
    positions = dpluspy.utils._read_bed_file_positions(bed_fname) + 1
    mut_map = dpluspy.parsing._load_mutation_map(mut_fname, positions)
    _numer = np.sum(mut_map)
    _denom = len(mut_map)
    numer += _numer 
    denom += _denom 
    print(f"chromosome {chrom} L={_denom} u={_numer / _denom}")

print(f"genome-wide L={denom} u={numer / denom}")
```
