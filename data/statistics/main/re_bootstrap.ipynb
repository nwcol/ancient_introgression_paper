```python
# save 510 bootstrap samples

def run_bootstrap(dataset):
    sums = {}
    for chrom in range(1, 23):
        with open(f"raw_sums/sums_{dataset}_weighted_chr_{chrom}.pkl", "rb") as fin:
            sums.update(pickle.load(fin))
    bins = sums[next(iter(sums))]["bins"]
    pop_ids = sums[next(iter(sums))]["pop_ids"]
    means, varcovs = dpluspy.bootstrapping.bootstrap_stats(sums, num_reps=510)
    bootreps = dpluspy.bootstrapping.get_bootstrap_reps(sums, num_reps=510)
    data = {
        "means": means, 
        "varcovs": varcovs,
        "bootreps": bootreps,
        "bins": bins,
        "pop_ids": pop_ids
    }
    with open(f"{dataset}_stats_510_samples.pkl", "wb") as fout:
        pickle.dump(data, fout)
    return

run_bootstrap("ZhouJHS")
run_bootstrap("Bherer")
```
