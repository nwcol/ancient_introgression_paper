```python
import numpy as np
import dpluspy
import pandas
```

```python
chroms = list(range(1, 23))
masks = [
    "filterbed",
    "filterbed_roulette",
    "filterbed_intergenic",
    "filterbed_10kb_exon_buffer",
    "filterbed_20kb_exon_buffer",
    "filterbed_50kb_exon_buffer",
    "filterbed_1e-4M_exon_buffer"
]
data = {x: [] for x in masks}
for mask in masks:
    for chrom in chroms:
        fname = f"{mask}/mask_chr{chrom}.bed.gz"
        regs, _ = dpluspy.utils._read_bed_file(fname)
        num_sites = np.sum(regs[:, 1] - regs[:, 0])
        data[mask].append(num_sites)
```

```python
out_data = {"chrom": [f"chr{c}" for c in chroms]}
out_data.update(data)
df = pandas.DataFrame(out_data)
df.to_csv("coverage_summary.csv", index=False)
```
