```python
import pandas
import numpy as np
```

```python
bands = pandas.read_csv("../../cytoBand.txt", sep="\s+",
    names=["chrom", "start", "end", "band", "stain"])
seqlens = pandas.read_csv("../../hg19.genome", sep="\s+",
    names=["chrom", "seqlen"])
```

```python
seqlens
```

```python
bands[bands["chrom"] == "chr19"]
```

```python
for ii in range(1, 23):
    chrom = f"chr{ii}"
    chrbands = bands[bands["chrom"] == chrom]
    seq_len = np.unique(seqlens[seqlens["chrom"] == chrom]["seqlen"])[0]
    if ii in (13, 14, 15, 21, 22):
        intervals = [[1, seq_len + 1]]
    else:
        centromere = chrbands[chrbands["stain"] == "acen"]
        breakpoint = (np.min(centromere["start"]) + np.max(centromere["end"])) / 2
        intervals = [[1, breakpoint], [breakpoint, seq_len + 1]]
    np.savetxt(f"arms_{chrom}.txt", np.array(intervals), fmt="%d")
```
