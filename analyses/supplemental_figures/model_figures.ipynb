```python
import numpy as np
import h2py
import pickle
import matplotlib.pyplot as plt
import matplotlib as mpl
import scipy
import pandas
import demes
import demesdraw


mpl.rc("xtick", labelsize=7)
mpl.rc("ytick", labelsize=7)
mpl.rc("font", size=8)
mpl.rc("axes", labelsize=8)
mpl.rc("axes", titlesize=8)
mpl.rc("legend", fontsize=7)
mpl.rc("axes", titlesize="small")
```

```python
graph0 = "../../models/main/submodels/bherer_amh_nea_no_pulse.yaml"
graph1 = "../../models/main/submodels/bherer_amh_nea_one_pulse.yaml"
graph2 = "../../models/main/submodels/bherer_amh_nea_two_pulse.yaml"
fname = "../../models/bootstrapped_data/accessible_1e-4M_ccds_flank_zhou-JHS_adj.pkl"
with open(fname, "rb") as fin:
    data = pickle.load(fin)
graph0 = demes.load(graph0)
deme_names = [d.name for d in graph0.demes]
pops = [p for p in data["pops"] if p in deme_names]
data = h2py.parsing.subset_data(data, to_pops=pops)
bins = h2py.parsing._get_bin_tuples(np.logspace(-6, -2, 17))
model0 = h2py.H2stats.from_demes(graph0, u=1.3e-8, r_bins=bins, sampled_demes=pops)
model1 = h2py.H2stats.from_demes(graph1, u=1.3e-8, r_bins=bins, sampled_demes=pops)
model2 = h2py.H2stats.from_demes(graph2, u=1.3e-8, r_bins=bins, sampled_demes=pops)
data = h2py.parsing.subset_data(data, to_pops=pops)
p1s = [x.split("_")[1] for x in data["stats"][0]]
p2s = [x.split("_")[2] for x in data["stats"][0]]
```

```python
p1s = [int(x.split("_")[1]) for x in data["stats"][0]]
p2s = [int(x.split("_")[2]) for x in data["stats"][0]]
stats = [f"{pops[p1][:3]}, {pops[p2][:3]}" for p1, p2 in zip(p1s, p2s)]

x = h2py.plotting.plot_h2_curves(
    h2_stats=[model0, model1, model2],
    means=[data["means"]],
    varcovs=[data["varcovs"]],
    colors=["tab:grey", "tab:blue", "tab:orange", "tab:green"],
    stats=stats,
    fig_size=(8.5, 5.5),
    pops=pops,
    rows=4, cols=6,
    r_bins=bins, #data["bins"],
    model_labels=["isolation", "one-pulse", "two-pulse"],
    data_labels=["data (Bhérer)"],
    out_file="../../figures/amh_neand_models.pdf"
)
```
