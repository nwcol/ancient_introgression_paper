```python
def plot_graph(ax):
    fname = "../models/full_model/round_1/fitted_models/full_model_round_0_Bherer_fit_20.yaml"
    positions = {
        "Yoruba1": -1e4,
        "Stuttgart": 3e4,
        "Loschbour": 6e4,
        "UstIshim": 9e4,
        "Vindija": 1.2e5,
        "Chagyrskaya": 1.5e5, 
        "Altai": 1.8e5,
        "Denisova": 2.1e5,
        "MH": 3e4, 
        "WN": 1.35e5,
        "N": 1.55e5,
        "ND": 1.81e5,
        "Anc": 1.05e5
    }
    # map deme names to desired labels
    labels = {
        "Yoruba1": "Yoruba", 
        "UstIshim": "Ust'Ishim", 
        "Loschbour": "Loschbour", 
        "Stuttgart": "Stuttgart", 
        "Vindija": "Vindija", 
        "Chagyrskaya": "Chagyrskaya", 
        "Altai": "Altai", 
        "Denisova": "Denisova"
    }
    sample_times = {
        "Yoruba1": 0, 
        "UstIshim": 45000, 
        "Loschbour": 8000, 
        "Stuttgart": 7000, 
        "Vindija": 45000, 
        "Chagyrskaya": 55000, 
        "Altai": 115000, 
        "Denisova": 60000
    }
    graph = demes.load(fname)
    demesdraw.tubes(
        graph,
        ax=ax,
        # colours="blue",
        labels=None,
        log_time=True,
        inf_ratio=0.02,
        positions=positions
    )
    ax.spines["top"].set_visible(True)
    ax.spines["right"].set_visible(True)
    ax.spines["bottom"].set_visible(True)
    ax.set_xticks([positions[x] for x in labels], [labels[x] for x in labels])
    #yticks = [5, 10, 50, 100, 500, 1000]
    #ax.set_yticks(np.array(yticks) * 1e3, yticks)
    #ax.set_ylabel("time (kya)")
    ax.set_ylabel("Time (years)")

    ax.annotate("Vindija", (positions["Vindija"] - 7e3, 3.9e4), fontsize="small")
    ax.annotate("Chagyrskaya", (positions["Chagyrskaya"] - 1e4, 4.8e4), fontsize="small")
    ax.annotate("Altai", (positions["Altai"] - 5e3, 9.5e4), fontsize="small")
    ax.annotate("Denisova", (positions["Denisova"] - 7e3, 5.2e4), fontsize="small")
    ax.annotate("Ust'Ishim", (positions["UstIshim"] - 1.1e4, 3.8e4), fontsize="small")
    ax.annotate("Loschbour", (positions["Loschbour"] - 1e4, 6e3), fontsize="small")
    ax.annotate("Stuttgart", (positions["Stuttgart"] - 1e4, 4.4e3), fontsize="small")
    ax.set_xticks([positions["Yoruba1"]], ["Yoruba"], fontsize="small")

    ax.set_ylim(4e3,)
    return
```

```python
mosaic = "A"
fig, axs = plt.subplot_mosaic(mosaic, figsize=(7.5, 6), layout="constrained")
for label, ax in axs.items():
    ax.text(0.0, 1.0, label, fontsize="large", va="bottom", transform=(
        ax.transAxes + ScaledTranslation(-20/72, +7/72, fig.dpi_scale_trans)))
plot_graph(axs["A"])
plt.savefig("figures/fig6.pdf")
```
