```python
import demes
import scipy
import dpluspy
import pandas 
import numpy as np
import moments
import matplotlib
import matplotlib.pyplot as plt

font_size = 8
matplotlib.rc("xtick", labelsize=font_size)
matplotlib.rc("ytick", labelsize=font_size)
matplotlib.rc("axes", labelsize=font_size)
matplotlib.rc("axes", titlesize=font_size)
matplotlib.rc("legend", fontsize=font_size)
```

```python
model_files = [
    "fitted_models/model_0_fit_3.yaml",
    "fitted_models/model_1_fit_3.yaml",
    "fitted_models/model_2_fit_24.yaml",
    "fitted_models/model_3_fit_21.yaml",
    "fitted_models/model_4_fit_22.yaml",
    "fitted_models/model_5_fit_23.yaml",
    "fitted_models/model_6_fit_6.yaml",
    "fitted_models/model_7_fit_0.yaml"
]
data_file = "../../../analyses/recombination_maps/stats/subset_Bherer.pkl"
u = 1.3e-8

data = dpluspy.inference.load_bootstrap_reps(
    data_file, graph=model_files[0], return_dict=True)
models = [
    dpluspy.inference.compute_bin_stats(file, sampled_demes=data["pop_ids"], 
        bins=data["bins"], u=u) for file in model_files
] 

```

```python
dpluspy.plotting.plot_D_plus_curves(
    models=models,
    pop_ids=data["pop_ids"], 
    bins=data["bins"], 
    means=data["means"], 
    varcovs=data["varcovs"], 
    out="figures/best_fit_models.pdf",
    labels=["data"] + [f"model {i}" for i in range(8)],
    ax_size=1.75
)
```
