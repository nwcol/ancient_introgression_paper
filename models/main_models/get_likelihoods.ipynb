```python
import os
import moments
import dpluspy
import pickle
import numpy as np
import pandas
```

```python
"""
create a table holding log-likelihoods for each model.
"""


models = [
    "amh_nea_no_pulse",
    "amh_nea_one_pulse",
    "amh_nea_two_pulse",
    "amh_nea_two_pulse_free_time",
    "sup_den_no_pulse",
    "sup_den",
    "sup_den_without_amh_nea_no_pulse",
    "sup_den_without_amh_nea",
    "sup_den_free_time",
    "los_stu_isolation",
    "los_stu_migration",
    "los_stu_admixture",
    "full",
    "full_fast_mutation",
    "full_slow_mutation",
    "full_without_sup_den",
    "full_edited"
]

data_sets = ["bherer", "zhou"]


entries = {"model": [], "ll": []}

for data_set in data_sets:
    for model in models:
        fname = f"{data_set}_{model}.yaml"
        if fname not in os.listdir():
            print(f"missing file {fname}")
            continue
        pop_ids, bins, data, varcovs = dpluspy.inference.load_stats(
            f"../../statistics/{data_set}_stats.pkl", graph=fname)
        exps = dpluspy.inference.compute_bin_stats(
            fname, pop_ids, u=1.3e-8, bins=bins)
        dpluspy.inference._inv_varcov_cache = {}
        ll = dpluspy.inference.composite_ll(exps, data, varcovs)
        entries["model"].append(f"{data_set}_{model}")
        entries["ll"].append(ll)
        
pandas.DataFrame(entries).to_csv("likelihood_tbl.csv", index=False)
```
