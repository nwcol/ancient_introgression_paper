```python
import numpy as np
import pandas 
import demes
import h2py
import pickle
```

```python
def printout(mapping, fixed_params, dfs):
    """
    Arguments
    ---------
    mapping : dict
        maps internal parameter name to tuple of (printout parameter name, scale)
    fixed_params : dict
        maps fixed parameter names to fixed values
    dfs : list of dataframes
    """
    for pname in mapping:
        fancyname, fac = mapping[pname]
        out = [fancyname]
        for df in dfs:
            if pname in list(df["name"]):
                mle = next(iter(df[df["name"] == pname]["MLE"])) * fac
                mle_str = f"{np.format_float_positional(mle, precision=3, fractional=False, trim='-')}"
                lower = next(iter(df[df["name"] == pname]["CI_lower"])) * fac
                upper = next(iter(df[df["name"] == pname]["CI_upper"])) * fac
                ci_str = (f"{np.format_float_positional(lower, precision=3, fractional=False, trim='-')} "
                        f"-- {np.format_float_positional(upper, precision=3, fractional=False, trim='-')}")
            elif pname in fixed_params:
                mle = fixed_params[pname] * fac
                mle_str = f"{np.format_float_positional(mle, precision=3, fractional=False, trim='-')}"
                ci_str = "---"
            else:
                mle_str = ""
                ci_str = ""
            out += [mle_str, ci_str]
        out = " & ".join(out) + r" \\"
        print(out)
    return
```

```python
df0 = pandas.read_csv("mutation_sensitivity/bootstrap_bherer_1.3e-8.csv")
df1 = pandas.read_csv("mutation_sensitivity/bootstrap_zhou-JHS_1.3e-8.csv")


mapping = {
    "T_S": (r"$T_{\text{NDH--S}}$ (ka) (fixed)", 1e-3),
    'T_NDH': (r'$T_{\text{ND--H}}$ (ka)', 1e-3),
    'T_ND': (r'$T_{\text{N--Den}}$ (ka)', 1e-3),
    "T_HN0": (r"$T_{\text{H} \to \text{N}}$ (ka) (fixed)", 1e-3), 
    'T_AWN': (r'$T_{\text{WN--Alt}}$ (ka)', 1e-3),
    "T_HN1": (r"$T_{\text{H} \to \text{WN}}$ (ka) (fixed)", 1e-3),
    'T_CV': (r'$T_{\text{Cha--Vin}}$ (ka) ', 1e-3),
    'T_OOA': (r'$T_{\text{Yor--OOA}}$ (ka)', 1e-3),
    'T_SL': (r'$T_{\text{OOA--Stu}}$ (ka)', 1e-3),
    "T_NH": (r"$T_{\text{Vin} \to \text{OOA}}$ (ka) (fixed)", 1e-3), 
    "T_Ust": (r"$T_{\text{OOA--Ust}}$ (ka) (fixed)", 1e-3), 
    'T_L_S': (r'$T_{\text{Los$\to$Stu}}$ (ka)', 1e-3),
    'N_Anc': (r'$N_{\text{A}}$', 1),
    "N_S": (r"$N_{\text{S}}$ (fixed)", 1),
    'N_ND': (r'$N_{\text{ND}}$', 1),
    'N_N': (r'$N_{\text{N}}$', 1),
    'N_MH': (r'$N_{\text{H}}$ ', 1),
    'N_OOA': (r'$N_{\text{OOA}}$', 1),
    'N_Alt': (r'$N_{\text{Alt}}$', 1),
    'N_Chag': (r'$N_{\text{Cha}}$', 1),
    'N_Vin': (r'$N_{\text{Vin}}$', 1),
    'N_Den': (r'$N_{\text{Den}}$', 1),
    'N_Los': (r'$N_{\text{Los}}$', 1),
    'N_Ust': (r'$N_{\text{Ust}}$', 1),
    'N_Stut': (r'$N_{\text{Stu}}$', 1),
    'N_Yor': (r'$N_{\text{Yor}}$', 1),
    'm_ND': (r'$m_{\text{N--Den}}$ ($\times 10^{-5}$)', 1e5),
    'm_LY': (r'$m_{\text{Yor--Los}}$ ($\times 10^{-5}$)', 1e5),
    'p_S_D': (r'$f_{\text{S} \to \text{D}}$', 1),
    'p_H_N': (r'$f_{\text{H} \to \text{N}}$', 1),
    'p_H_WN': (r'$f_{\text{H} \to \text{WN}}$', 1),
    'p_N_OOA': (r'$f_{\text{Vin}\to\text{OOA}}$', 1),
    'p_L_S': (r'$f_{\text{Los}\to\text{Stu}}$', 1)
}

fixed_params = {
    "T_S": 2e6,
    "T_HN0": 2.5e5, 
    "T_HN1": 1.1e5, 
    "T_NH": 48000, 
    "T_Ust": 47000, 
    "N_S": 20000, 
}


printout(mapping, fixed_params, [df0, df1])
```

```python
data_fname = "bootstrapped_data/accessible_1e-4M_ccds_flank_bherer_adj.pkl"
graph0 = "../all_models/main_models/bherer_amh_nea_no_pulse.yaml"
graph1 = "../all_models/main_models/bherer_amh_nea_one_pulse.yaml"
lrt_fname = "bherer_amh_n_pulse.csv"

with open(data_fname, "rb") as fin:
    data = pickle.load(fin)
graph = demes.load(graph0)
deme_names = [d.name for d in graph.demes]
pops = [p for p in data["pops"] if p in deme_names]
data = h2py.parsing.subset_data(data, to_pops=pops)
bins = h2py.parsing._get_bin_tuples(np.logspace(-6, -2, 17))
model0 = h2py.H2stats.from_demes(graph0, u=1.3e-8, r_bins=bins, sampled_demes=pops)
model1 = h2py.H2stats.from_demes(graph1, u=1.3e-8, r_bins=bins, sampled_demes=pops)
ll0 = h2py.inference._compute_composite_ll(model0, data["means"], data["varcovs"])
ll1 = h2py.inference._compute_composite_ll(model1, data["means"], data["varcovs"])
print(ll0, ll1)
df = pandas.read_csv(f"../all_models/blrt/{fname}")
scores = np.array(df["ll1"]) - np.array(df["ll0"])
print(ll1 - ll0, np.count_nonzero(scores >= (ll1 -ll0)))
```
