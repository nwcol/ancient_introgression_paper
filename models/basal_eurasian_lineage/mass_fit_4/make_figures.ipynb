```python
import dpluspy
import pandas 
import numpy as np
from collections import defaultdict
```

```python
u = 1.3e-8
Bhe_data_file = "../../../analyses/recombination_maps/stats/subset_Bherer.pkl"
Zho_data_file = "../../../analyses/recombination_maps/stats/subset_ZhouJHS.pkl"
```

```python
# 5-deme Bherer models

model_Bhe0_5_file = "fitted_models/Bherer/model_0_5_Bherer_fit_0.yaml"
model_Bhe1_5_file = "fitted_models/Bherer/model_1_5_Bherer_fit_3.yaml"
model_Bhe2_5_file = "fitted_models/Bherer/model_2_5_Bherer_fit_17.yaml"
model_Bhe3_5_file = "fitted_models/Bherer/model_3_5_Bherer_fit_24.yaml"

data_Bhe_5 = dpluspy.inference.load_bootstrap_reps(
    Bhe_data_file, graph=model_Bhe0_5_file, return_dict=True)

model_Bhe0_5 = dpluspy.inference.compute_bin_stats(model_Bhe0_5_file, 
    sampled_demes=data_Bhe_5["pop_ids"], bins=data_Bhe_5["bins"], u=u)
model_Bhe1_5 = dpluspy.inference.compute_bin_stats(model_Bhe1_5_file, 
    sampled_demes=data_Bhe_5["pop_ids"], bins=data_Bhe_5["bins"], u=u)
model_Bhe2_5 = dpluspy.inference.compute_bin_stats(model_Bhe2_5_file, 
    sampled_demes=data_Bhe_5["pop_ids"], bins=data_Bhe_5["bins"], u=u)
model_Bhe3_5 = dpluspy.inference.compute_bin_stats(model_Bhe3_5_file, 
    sampled_demes=data_Bhe_5["pop_ids"], bins=data_Bhe_5["bins"], u=u)
```

```python
dpluspy.plotting.plot_D_plus_curves(
    models=[model_Bhe0_5, model_Bhe1_5, model_Bhe2_5, model_Bhe3_5],
    pop_ids=data_Bhe_5["pop_ids"], 
    bins=data_Bhe_5["bins"], 
    means=data_Bhe_5["means"], 
    varcovs=data_Bhe_5["varcovs"], 
    title="5-deme models, Bherer recombination map",
    out="figures/figure_curves_Bherer_5_deme_models.pdf",
    labels=[
        "data", 
        "model 0 [tree-OOA]", 
        "model 1 [basal eurasian]", 
        "model 2 [deep structure]", 
        "model 3 [shallow structure]"
    ],
    ax_size=1.75,
    cols=5
)
```
