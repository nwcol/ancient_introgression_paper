```python
import numpy as np
import pandas
import dpluspy 
import demes
import demesdraw
import matplotlib, matplotlib.pylab as plt

matplotlib.rc("xtick", labelsize=8)
matplotlib.rc("ytick", labelsize=8)
matplotlib.rc("axes", labelsize=8)
matplotlib.rc("axes", titlesize=8)
matplotlib.rc("legend", fontsize=8)
```

```python

u = 1.3e-8

graph = "fitted_models/deep_struc_model_fit_28.yaml"
data_file = "../../../analyses/recombination_maps/stats/subset_Bherer.pkl"

data = dpluspy.inference.load_bootstrap_reps(
    data_file, graph=graph, return_dict=True)

model = dpluspy.inference.compute_bin_stats(
    graph, sampled_demes=data["pop_ids"], bins=data["bins"], u=u)
```

```python
dpluspy.plotting.plot_D_plus_curves(
    models=model,
    pop_ids=data["pop_ids"], 
    bins=data["bins"], 
    means=data["means"], 
    varcovs=data["varcovs"], 
    ax_size=1.5,
    cols=7,
    out="figures/deep_struc_fit_14.pdf",    
)
```

```python
fig, ax = plt.subplots(figsize=(7.5, 5), layout="constrained")
demesdraw.tubes(demes.load(graph), ax=ax, log_time=True, inf_ratio=0.05,
        )
ax.set_ylim(4e3,)
plt.savefig("figures/deep_struc_fit_14_graph.pdf")
```

```python
# Make a table of fitted parameters

dfs = []
param_file = "dummy_params.yaml"

for rep in range(30):
    graph_file = f"fitted_models/deep_struc_model_fit_{rep}.yaml"
    pnames, params = dpluspy.inference._load_params(graph_file, param_file)
    pdict = {"rep": rep}
    pdict.update({pname: [p] for pname, p in zip(pnames, params)})
    dfs.append(pandas.DataFrame(pdict))

param_df = pandas.concat(dfs)
param_df.to_csv("fitted_parameters.csv", index=False)
print(param_df.to_string(index=False))
```

```python
# Comparison to two null models

graph_0 = "local_null_model_fit/basal_eurasian_model_fit.yaml"
graph_1 = "local_null_model_fit/null_model_local_fit.yaml"

model_0 = dpluspy.inference.compute_bin_stats(
    graph_0, sampled_demes=data["pop_ids"], bins=data["bins"], u=u)
model_1 = dpluspy.inference.compute_bin_stats(
    graph_1, sampled_demes=data["pop_ids"], bins=data["bins"], u=u)
```

```python
ll_0 = dpluspy.inference.composite_ll(model_0, data["means"], data["varcovs"])
print(f"LL_0 old fit = {ll_0}")
ll_1 = dpluspy.inference.composite_ll(model_1, data["means"], data["varcovs"])
print(f"LL_0 new fit = {ll_1}")
ll = dpluspy.inference.composite_ll(model, data["means"], data["varcovs"])
print(f"LL_0 structure fit = {ll}")
```

```python
dpluspy.plotting.plot_D_plus_curves(
    models=[model_0, model],
    pop_ids=data["pop_ids"], 
    bins=data["bins"], 
    means=data["means"], 
    varcovs=data["varcovs"], 
    ax_size=1.5,
    out="figures/model_comparisons.pdf",
    labels=[
        "data",
        "prior basal Eur. model",
        "weak stem model"
    ]
)
```
