```python
import collections
import demes
import moments 
import numpy as np
import pandas
```

```python
def load_best_tries(options_fname, prefix, n_tries=3, n_samples=510):
    """
    Load the best-fitting try for each bootstrap sample and concatenate 
    these together into an array.
    """
    options = moments.Demes.Inference._get_params_dict(options_fname)
    data = collections.defaultdict(list)
    for ii in range(n_samples):
        tries = []
        lls = []
        for jj in range(n_tries):
            graph_file = f"{prefix}sample_{ii}_try_{jj}.yaml"
            try:
                lls.append(demes.load(graph_file).metadata["opt_info"]["ll"])
                tries.append(graph_file)
            except:
                print(f"missing sample {ii} try {jj}")
        if len(tries) == 0:
            raise ValueError(f"missing input for sample {ii}")
        best_try = tries[np.argmax(lls)]
        builder = moments.Demes.Inference._get_demes_dict(best_try)
        pnames, params, *_  = moments.Demes.Inference._set_up_params_and_bounds(
            options, builder)
        data["sample"].append(ii)
        data["ll"].append(np.max(lls))
        for pname, param in zip(pnames, params):
            data[pname].append(param)
    return pandas.DataFrame(data)
```

```python
# full model variants

bherer_full = load_best_tries(
    "../main_models/params_full.yaml", "bherer_full/", n_tries=3)
bherer_full.to_csv("bherer_full.csv", index=False)

zhou_full = load_best_tries(
    "../main_models/params_full.yaml", "zhou_full/", n_tries=3)
zhou_full.to_csv("zhou_full.csv", index=False)


bherer_full_without_sup_den = load_best_tries(
    "../main_models/params_full_without_sup_den.yaml", 
    "bherer_full_without_sup_den/", n_tries=3)
bherer_full_without_sup_den.to_csv(
    "bherer_full_without_sup_den.csv", index=False)

zhou_full_without_sup_den = load_best_tries(
    "../main_models/params_full_without_sup_den.yaml", 
    "zhou_full_without_sup_den/", n_tries=3)
zhou_full_without_sup_den.to_csv(
    "zhou_full_without_sup_den.csv", index=False)


bherer_full_slow_mutation = load_best_tries(
    "../main_models/params_full.yaml", 
    "bherer_full_slow_mutation/", n_tries=2)
bherer_full_slow_mutation.to_csv(
    "bherer_full_slow_mutation.csv", index=False)


bherer_full_fast_mutation = load_best_tries(
    "../main_models/params_full.yaml", 
    "bherer_full_fast_mutation/", n_tries=2)
bherer_full_fast_mutation.to_csv(
    "bherer_full_fast_mutation.csv", index=False)
```
