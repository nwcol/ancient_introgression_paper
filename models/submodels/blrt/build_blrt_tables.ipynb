```python
import numpy as np
import pandas 
```

```python
def load_best_tries(directory, n_samples=500, n_tries=5):
    lls_null = []
    lls_alt = []
    lrt_scores = []
    for ii in range(n_samples):
        sample_lls_null = []
        sample_lls_alt = []
        for jj in range(n_tries):
            csv_file = f"{directory}/sample_{ii}_try_{jj}.csv"
            try:
                ll_null, ll_alt = list(pandas.read_csv(csv_file)["ll"])
                sample_lls_null.append(ll_null)
                sample_lls_alt.append(ll_alt)
            except:
                print(f"missing sample {ii} try {jj}")
        if len(sample_lls_null) == 0:
            max_ll_null = np.nan
            max_ll_alt = np.nan
            lrt_score = np.nan
            print(f"missing sample {ii}")
        else:
            max_ll_null = np.max(sample_lls_null)
            max_ll_alt = np.max(sample_lls_alt)
            lrt_score = 2 * (max_ll_alt - max_ll_null)
        lrt_scores.append(lrt_score)
        lls_null.append(max_ll_null)
        lls_alt.append(max_ll_alt)
    lrt_scores = np.array(lrt_scores)
    lrt_scores[lrt_scores < 0] = 0
    data = {"sample": list(range(n_samples)), "ll0": lls_null, "ll1": lls_alt, "score": lrt_scores}
    return pandas.DataFrame(data)
```

```python
bherer_amh_n_pulse = load_best_tries("bherer_amh_n_pulse")
bherer_amh_n_pulse.to_csv("bherer_amh_n_pulse.csv", index=False)

zhou_amh_n_pulse = load_best_tries("zhou_amh_n_pulse")
zhou_amh_n_pulse.to_csv("zhou_amh_n_pulse.csv", index=False)


bherer_amh_wn_pulse = load_best_tries("bherer_amh_wn_pulse")
bherer_amh_wn_pulse.to_csv("bherer_amh_wn_pulse.csv", index=False)

zhou_amh_wn_pulse = load_best_tries("zhou_amh_wn_pulse")
zhou_amh_wn_pulse.to_csv("zhou_amh_wn_pulse.csv", index=False)


bherer_sup_den_pulse = load_best_tries("bherer_sup_den_pulse")
bherer_sup_den_pulse.to_csv("bherer_sup_den_pulse.csv", index=False)

zhou_sup_den_pulse = load_best_tries("zhou_sup_den_pulse")
zhou_sup_den_pulse.to_csv("zhou_sup_den_pulse.csv", index=False)


bherer_sup_den_pulse_wthout_amh_nea = load_best_tries("bherer_sup_den_pulse_without_amh_nea")
bherer_sup_den_pulse_wthout_amh_nea.to_csv("bherer_sup_den_pulse_without_amh_nea.csv", index=False)

zhou_sup_den_pulse_wthout_amh_nea = load_best_tries("zhou_sup_den_pulse_without_amh_nea")
zhou_sup_den_pulse_wthout_amh_nea.to_csv("zhou_sup_den_pulse_without_amh_nea.csv", index=False)


bherer_los_stu_pulse = load_best_tries("bherer_los_stu_pulse")
bherer_los_stu_pulse.to_csv("bherer_los_stu_pulse.csv", index=False)

zhou_los_stu_pulse = load_best_tries("zhou_los_stu_pulse")
zhou_los_stu_pulse.to_csv("zhou_los_stu_pulse.csv", index=False)
```
