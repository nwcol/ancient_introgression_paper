```python
import gzip
import dpluspy
import numpy as np
import pickle
```

```python
with open("bherer_naive_raw_sums.pkl", "rb") as fin:
    full_dataset = pickle.load(fin)


def run_bootstrap(dataset, output):
    bins = dataset[next(iter(dataset))]["bins"]
    pop_ids = dataset[next(iter(dataset))]["pop_ids"]
    means, varcovs = dpluspy.bootstrapping.bootstrap_stats(dataset, num_reps=100)
    data = {
        "means": means, 
        "varcovs": varcovs,
        "bins": bins,
        "pop_ids": pop_ids
    }
    with open(output, "wb") as fout:
        pickle.dump(data, fout)
    return


for chrom in range(1, 23):
    subset = {key: full_dataset[key] for key in full_dataset
              if key[0] == chrom}
    output = f"chromosome_stats/bherer_naive_chr{chrom}.pkl"
    run_bootstrap(subset, output)
```
