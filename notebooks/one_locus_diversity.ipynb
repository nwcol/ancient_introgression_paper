Let ni, Ni be the count of heterozygous sites and loci in window i. Here, note that H^2 is approximately equal to D+(total) across some contiguous genomic region composed of windows 1, 2 and 3.

```python
# define k choose 2 function
bn = lambda k: k * (k - 1) // 2

n1 = 500
n2 = 1350
n3 = 1500
N1 = 5e5
N2 = N3 = 1e6
H1 = n1 / N1
H2 = n2 / N2
H3 = n3 / N3
H_tot = (n1 + n2 + n3) / (N1 + N2 + N3)
Dp_tot = bn(n1 + n2 + n3) / bn(N1 + N2 + N3)
print(f"D+_tot = {Dp_tot}")
print(f"H_tot = {H_tot}")
print(f"H_tot^2 = {H_tot ** 2}")
```

Now note also that D+(total) measured only *within* windows corresponds to an H^2 statistic composed as a sum of squares. This in turn is equivalent to a statistic composed by weighting window H estimates by the number of locus pairs in each window.

```python
Dp_within = (bn(n1) + bn(n2) + bn(n3)) / (bn(N1) + bn(N2) + bn(N3))
Hsqr_within = (n1 ** 2 + n2 ** 2 + n3 ** 2) / (N1 ** 2 + N2 ** 2 + N3 ** 2)
Hsqr_weight = (H1 ** 2 * bn(N1) + H2 ** 2 * bn(N2) + H3 ** 2 * bn(N3)) / (bn(N1) + bn(N2) + bn(N3))
print(f"D+_within = {Dp_within}")
print(f"H^2_within = {Hsqr_within}")
print(f"H^2_weight = {Hsqr_weight}")
```

Now consider the case where windows 1 and 2 are adjacent while window 3 falls on a different chromosome, so that we have a count of pairs T1 for window 1 which includes pairs formed within window 1 and between windows 1 and 2.

```python
T1 = bn(N1) + N1 * N2 
t1 = bn(n1) + n1 * n2
Dp_tot = (t1 + bn(n2) + bn(n3)) / (T1 + bn(N2) + bn(N2))
Hsqr_sqr = ((n1 + n2) ** 2 + n3 ** 2) / ((N1 + N2) ** 2 + N3 ** 2)
H12 = (H1 * N1 + H2 * N2) / (N1 + N2)
Hsqr_weight = (H12 ** 2 * (T1 + bn(N2)) + H3 ** 2 * bn(N3)) / (T1 + bn(N2) + bn(N3))
print(f"D+_tot = {Dp_tot}")
print(f"H^2 = {Hsqr_sqr}")
print(f"H^2_weight = {Hsqr_weight}")
print(f"root H^2_weight = {Hsqr_weight ** 0.5}")
```
