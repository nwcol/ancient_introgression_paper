```python
import pandas
import numpy as np
```

```python
models = [
    "000",
    "100",
    "010",
    "001",
    "110",
    "101",
    "011",
    "111"
]
nreps = 500

for model in models:
    tbls = []
    for ii in range(nreps):
        fname = f"fits/rep_{ii}_model_{model}.csv"
        tbl = pandas.read_csv(fname)
        tbls.append(tbl)
    output_tbl = pandas.concat(tbls)
    output_tbl.to_csv(f"fits_model_{model}.csv", index=0)
```
