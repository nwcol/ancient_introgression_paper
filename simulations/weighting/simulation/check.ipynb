```python
import numpy as np
import demes 
import dpluspy 
import matplotlib.pyplot as plt
```

```python
data0 = dpluspy.inference.load_stats("test/heterog_u_1_rep_0_naive_stats.pkl",
                                    return_dict=True)
data1 = dpluspy.inference.load_stats("test/heterog_u_1_rep_0_weighted_stats.pkl",
                                    return_dict=True)
graph = "full_model.yaml"
u = 1.152519e-08
model = dpluspy.inference.compute_bin_stats(graph, u=u, 
    sampled_demes=data0["pop_ids"], bins=data0["bins"])
dpluspy.plotting.plot_D_plus_curves(
    models=[model], 
    means=[data0["means"], data1["means"]], 
    varcovs=[data0["varcovs"], data1["varcovs"]],
    bins=data0["bins"], pop_ids=data0["pop_ids"]
)
```
