```python
# basic check of correspondence between realizations/expectations

import dpluspy
import demes
import numpy as np
import pandas
import pickle
```

```python
data = [dpluspy.inference.load_stats(
    f"trials/hn_null_model_rep_{i}.pkl", return_dict=True) for i in range(5)]
model = dpluspy.inference.compute_bin_stats("../models/hn_null_model.yaml", 
    sampled_demes=data[0]["pop_ids"], u=1.3e-08, bins=data[0]["bins"])

dpluspy.plotting.plot_D_plus_curves(
    models=[model],
    means=[x["means"] for x in data],
    varcovs=[x["varcovs"] for x in data],
    bins=data[0]["bins"],
    pop_ids=data[0]["pop_ids"],
    cols=3,
    ax_size=1.75,
    colors=["tab:orange"] * len(data) + ["black"]
)
```
