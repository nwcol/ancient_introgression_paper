```python
import numpy as np
import pandas 
import matplotlib.pyplot as plt
```

```python
# load samples

models = [
    "hn_null_model",
    "hn_model_1",
    "hn_model_5",
    "sd_null_model",
    "sd_model_1",
    "sd_model_5"
]

boot_results = {}

for model in models:
    for rep in range(10):
        ll0s = []
        ll1s = []
        samples = []
        for sample in range(100):
            fname = f"replicate_fits/results/{model}_rep_{rep}_samp_{sample}_mle.csv"
            try:
                tbl = pandas.read_csv(fname)
                ll0, ll1 = list(tbl["ll"])
                ll0s.append(ll0)
                ll1s.append(ll1)
                samples.append(sample)
            except:
                pass
        scores = 2 * (np.array(ll1s) - np.array(ll0s))
        scores[scores < 0] = 0
        data = {
            "sample": samples,
            "ll0": ll0s,
            "ll1": ll1s,
            "score": scores
        }
        pandas.DataFrame(data).to_csv(
            f"results/{model}_rep_{rep}_samples.csv", index=False)
        boot_results[f"{model}_rep_{rep}"] = scores
```

```python
# what fraction of samples have negative scores (due to convergence issues?)

models = [
    "hn_null_model",
    "hn_model_1",
    "hn_model_5",
    "sd_null_model",
    "sd_model_1",
    "sd_model_5"
]

for model in models:
    for rep in range(10):
        ll0s = []
        ll1s = []
        for sample in range(100):
            fname = f"replicate_fits/results/{model}_rep_{rep}_samp_{sample}_mle.csv"
            try:
                tbl = pandas.read_csv(fname)
                ll0, ll1 = list(tbl["ll"])
                ll0s.append(ll0)
                ll1s.append(ll1)
            except:
                pass
        scores = 2 * (np.array(ll1s) - np.array(ll0s))
        print(f"{model}_rep_{rep}", np.count_nonzero(scores < 0) / len(scores))
```

```python
# load whole-data results

whole_results = {"trial": [], "ll0": [], "ll1": [], "score": []}

for model in models:
    for rep in range(10):
        fname = f"data_set_fits/trials/{model}_rep_{rep}_fit.csv"
        tbl = pandas.read_csv(fname)
        ll0, ll1 = list(tbl["ll"])
        score = 2 * (ll1 - ll0)
        if score < 0:
            score = 0
        whole_results["trial"].append(f"{model}_rep_{rep}")
        whole_results["ll0"].append(ll0)
        whole_results["ll1"].append(ll1)
        whole_results["score"].append(score)

whole_results = pandas.DataFrame(whole_results)
whole_results.to_csv("results/data_set_results.csv", index=False)
```

```python
title = r"Test for H$\to$N introgression $\mid$ $f_{H\to N} = 0$"
model = "hn_null_model"

fig, axs = plt.subplots(2, 5, figsize=(12, 5), layout="constrained")
fig.suptitle(title)

for rep, ax in enumerate(axs.flat):
    samples = boot_results[f"{model}_rep_{rep}"]
    score = list(whole_results[whole_results["trial"] == f"{model}_rep_{rep}"]["score"])[0]
    p_obs = np.count_nonzero(samples > score) / len(samples)
    ax.hist(samples, bins=30, histtype="step", lw=1.3, density=1)
    ax.vlines(score, 0, 1, lw=1.3,
        transform=ax.get_xaxis_transform(), color="tab:orange")
    ax.set_title(f"$p_{{obs}}=${np.round(p_obs, 3)}")
    ax.set_ylabel("density")
    ax.set_xlabel("LRT score")
plt.savefig(f"{model}_results.pdf")
```
