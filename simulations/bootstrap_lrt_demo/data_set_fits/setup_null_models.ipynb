```python
import numpy as np
import pandas 
import dpluspy 
import demes
import moments
```

```python
null_model_file = "../models/hn_null_model.yaml"
null_param_file = "../models/hn_null_model_params.yaml"

builder = moments.Demes.Inference._get_demes_dict(null_model_file)
options = moments.Demes.Inference._get_params_dict(null_param_file)
pnames = [x["name"] for x in options['parameters']]

for model in ["null_model", "model_1", "model_5"]:
    print(model)
    for rep in range(10):
        fname = f"trials/hn_{model}_rep_{rep}_fit.csv"
        tbl = pandas.read_csv(fname)
        ll0, ll1 = list(tbl["ll"])
        print(ll1 - ll0)
        params = [list(tbl[pname])[0] for pname in pnames]
        builder = moments.Demes.Inference._update_builder(builder, options, params)
        graph = demes.Graph.fromdict(builder)
        out_fname = f"null_models/hn_{model}_rep_{rep}.yaml"
        demes.dump(graph, out_fname)
        
```
