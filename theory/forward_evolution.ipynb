```python
import numpy as np
import scipy
import moments
```

```python
# mutation
u = 1e-8
U = np.array(
    [[0, 0, 0],
     [1 * u, 0, 0],
     [0, 4 * u, 0]],
)
y = np.array([1, 0, 0])
for i in range(5):
    y = y + U @ y
    print(i, y)
```

```python
N = 5e3
x = -1 / (2 * N)
D = np.array(
    [[0, 0, 0],
     [0, x, 0],
     [0, 0, x]]
)
```

```python
y = np.array([1, 0, 0])
for i in range(100000):
    y = y + (D + U) @ y
print(y[1:])
```
